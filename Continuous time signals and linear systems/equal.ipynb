```python
from sympy import *
s, t, omega, n = symbols("s t omega n", real=True)
R1, R2, C1, C2, L1, L2 = symbols("R1 R2 C1 C2 L1 L2", positive=True, real=True)
VA, VB, V1 = symbols("VA VB V1")

num_a = L1*s/(C1*L1*R1*s**2 + R1 + R2 + s*(C1*R1*R2 + L1))
den_a = 1

num_b = 1/(R1*C1) * s
den_b = s**2 + (R2/L1 + 1/(R1*C1))*s + (1 + R2/R1)/(L1*C1)

a = num_a / den_a
b = num_b / den_b

print("a")
display(num_a, den_a,)
print("b")
display(num_b, den_b)

a = a.normal()
a = a.simplify()
a = a.expand()

b = b.normal()
b = b.simplify()
b = b.expand()

equal = a == b

print("equal?")
display(equal)
```
